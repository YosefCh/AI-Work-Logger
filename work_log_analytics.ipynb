# 🔍 Section 1: Query Work Logs

Search and filter your work log data using SQL or natural language

## 🔧 Direct SQL Query

<details> <summary style="color:rgb(19, 206, 134)"><strong>Click to Expand Details</strong></summary>

**Available Columns:**
- `Date` - Date of the work log (MM/DD/YYYY format)
- `Full_Text` - Complete work log entry  
- `Abridged` - Keywords/summary of the log

**Example Queries:**
```sql
SELECT * FROM work_logs WHERE Full_Text ILIKE '%Maestro%'
SELECT * FROM work_logs WHERE Date >= '2026-01-05'
SELECT Date, Abridged FROM work_logs WHERE Full_Text ILIKE '%Plex%' OR Full_Text ILIKE '%Fabric%'
```

</details>

````python
# Enter your SQL query here
sql_query = """
SELECT * FROM work_logs WHERE Abridged ILIKE '%plex%'
"""

# Execute the query
query_tool.run_sql_query(sql_query)
````
